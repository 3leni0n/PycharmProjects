```python
%matplotlib inline
import numpy as np
import scipy.integrate
from matplotlib import pyplot as plt
```

### Define activation function

```python
def F(x):
    return 1/(1+np.exp(-beta*(x-gamma)))

beta=4.0
gamma=0.7


```

### Define dynamics as before

```python
def IntegrateAuto(X, t):
    dXdT=-X+np.dot(J, F(X))+I
    return dXdT

def LookActAuto(_t,_x0):
    _x=scipy.integrate.odeint(IntegrateAuto,_x0,_t)
    return _x
```

### Define connectivity matrix

```python
J = np.zeros((2,2))
J_inh = -1

J[0,1] = J_inh
J[1,0] = J_inh
```

### Define nullclines
(!! this only works because we have no self-interactions)

```python
def nullclines_rhs(x):
    X=np.array([x,x])
    return np.dot(J, F(X))+I
```
